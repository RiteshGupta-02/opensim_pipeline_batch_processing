```python
import re

def extract_failures(log_path):
    current_subject = None
    current_trial = None
    failures = []

    subject_trial_re = re.compile(r'\[INFO\s+\].*?([Ss]\d+): Processing trial\s+(\w+)')
    # may also match [ERROR] lines that already include Sxx and trial text:
    err_with_trial_re = re.compile(r'\[ERROR\s+\].*?([Ss]\d+).*?(stw\d+)')
    optimization_re = re.compile(r'OPTIMIZATION FAILED|Optimizer failed|ID failed|SO failed|IK failed', re.IGNORECASE)

    with open(log_path, 'r', encoding='utf-8', errors='replace') as f:
        for line in f:
            m = subject_trial_re.search(line)
            if m:
                current_subject = m.group(1)
                current_trial = m.group(2)

            m2 = err_with_trial_re.search(line)
            if m2:
                current_subject = m2.group(1)
                current_trial = m2.group(2)

            if optimization_re.search(line):
                failures.append({
                    "subject": current_subject,
                    "trial": current_trial,
                    "line": line.strip()
                })

    return failures

if __name__ == "__main__":
    failures = extract_failures(r"d:\student\MTech\opensim_pipeline_batch_processing\pipeline\pipeline_4.txt")
    subject = None
    trial = None
    for f in failures:
        if  not f['subject'] == subject:
            subject = f['subject']
            trial = f['trial']
            print(f"{f['subject']} {f['trial']}  => {f['line']}")
```
